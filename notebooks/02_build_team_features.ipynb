## Aggregate season features
This section aggregates the long-format game logs into one row per team per season.

The goal is to summarize each team's regular season performance using average game-level statistics and season-level win/loss information.


```python
SEASON_AGG_DICT = {
    "Win": ["sum", "count"],
    "PointsFor": "mean",
    "PointsAgainst": "mean",
    "MarginScore": ["mean", "std"],
    "NumOT": "mean",
    "FieldGoalsMade": "mean",
    "FieldGoalsAttempted": "mean",
    "FieldGoalsPct": "mean",
    "ThreePointsMade": "mean",
    "ThreePointsAttempted": "mean",
    "ThreePointsPct": "mean",
    "FreeThrowsMade": "mean",
    "FreeThrowsAttempted": "mean",
    "FreeThrowPct": "mean",
    "OffensiveRebounds": "mean",
    "DefensiveRebounds": "mean",
    "TotalRebounds": "mean",
    "ReboundPct": "mean",
    "OffensiveReboundPct": "mean",
    "DefensiveReboundPct": "mean",
    "Assists": "mean",
    "Turnovers": "mean",
    "TurnoverPct": "mean",
    "AssistTurnoverRatio": "mean",
    "Steals": "mean",
    "Blocks": "mean",
    "PersonalFouls": "mean",
    "TurnoverMargin": "mean",
    "ReboundMargin": "mean",
    "Possessions": "mean",
    "OffEfficiency": "mean",
    "DefEfficiency": "mean",
    "NetRating": "mean",
    "GameTotalPoints": "mean",
}
```

```python
SEASON_FEATURE_COLUMNS = [
    "Season",
    "TeamID",
    "Wins",
    "GamesPlayed",
    "AvgPointsFor",
    "AvgPointsAgainst",
    "AvgMarginScore",
    "MarginScoreStd",
    "AvgNumOT",
    "AvgFieldGoalsMade",
    "AvgFieldGoalsAttempted",
    "AvgFieldGoalsPct",
    "AvgThreePointsMade",
    "AvgThreePointsAttempted",
    "AvgThreePointsPct",
    "AvgFreeThrowsMade",
    "AvgFreeThrowsAttempted",
    "AvgFreeThrowPct",
    "AvgOffensiveRebounds",
    "AvgDefensiveRebounds",
    "AvgTotalRebounds",
    "AvgReboundPct",
    "AvgOffensiveReboundPct",
    "AvgDefensiveReboundPct",
    "AvgAssists",
    "AvgTurnovers",
    "AvgTurnoverPct",
    "AvgAssistTurnoverRto",
    "AvgSteals",
    "AvgBlocks",
    "AvgPersonalFouls",
    "AvgTurnoverMargin",
    "AvgReboundMargin",
    "AvgPossessions",
    "AvgOffEfficiency",
    "AvgDefEfficiency",
    "AvgNetRating",
    "AvgGameTotalPoints",
]
```

```python
def build_team_season_features(games_df):
    team_features = games_df.groupby(
        ["Season", "TeamID"], as_index=False
    ).agg(SEASON_AGG_DICT)

    team_features.columns = SEASON_FEATURE_COLUMNS

    team_features["Losses"] = team_features["GamesPlayed"] - team_features["Wins"]
    team_features["WinPct"] = team_features["Wins"] / team_features["GamesPlayed"]
    team_features["LossPct"] = team_features["Losses"] / team_features["GamesPlayed"]

    return team_features
```

```python
team_features_men = build_team_season_features(games_men)
team_features_women = build_team_season_features(games_women)
```

## Conference features

```python
m_conf = pd.read_csv("../data/MTeamConferences.csv")
w_conf = pd.read_csv("../data/WTeamConferences.csv")
```

```python
team_features_men = team_features_men.merge(
    m_conf[["Season", "TeamID", "ConfAbbrev"]],
    on=["Season", "TeamID"],
    how="left"
)

team_features_women = team_features_women.merge(
    w_conf[["Season", "TeamID", "ConfAbbrev"]],
    on=["Season", "TeamID"],
    how="left"
)
```

## Seeds

```python
m_seeds = pd.read_csv("../data/MNCAATourneySeeds.csv")
w_seeds = pd.read_csv("../data/WNCAATourneySeeds.csv")
```

```python
def extract_seed_number(seed_value):
    if pd.isna(seed_value):
        return np.nan

    digits = "".join(ch for ch in str(seed_value) if ch.isdigit())
    return int(digits) if digits else np.nan
```

```python
m_seeds["SeedNum"] = m_seeds["Seed"].apply(extract_seed_number)
m_seeds["HasTournamentSeed"] = 1
```

```python
w_seeds["SeedNum"] = w_seeds["Seed"].apply(extract_seed_number)
w_seeds["HasTournamentSeed"] = 1
```

```python
team_features_men = team_features_men.merge(
    m_seeds[["Season", "TeamID", "Seed", "SeedNum", "HasTournamentSeed"]],
    on=["Season", "TeamID"],
    how="left"
)
team_features_men["HasTournamentSeed"] = (
    team_features_men["HasTournamentSeed"].fillna(0).astype(int)
)

team_features_women = team_features_women.merge(
    w_seeds[["Season", "TeamID", "Seed", "SeedNum", "HasTournamentSeed"]],
    on=["Season", "TeamID"],
    how="left"
)
team_features_women["HasTournamentSeed"] = (
    team_features_women["HasTournamentSeed"].fillna(0).astype(int)
)
```

## Massey

```python
m_massey = pd.read_csv("../data/MMasseyOrdinals.csv")
```

```python
massey_men = m_massey[m_massey["SystemName"] == "MAS"].copy()
```

```python
latest_massey_men = (
    massey_men
    .sort_values(["Season", "TeamID", "RankingDayNum"])
    .groupby(["Season", "TeamID"], as_index=False)
    .tail(1)
)

latest_massey_men = latest_massey_men[
    ["Season", "TeamID", "OrdinalRank"]
].rename(columns={
    "OrdinalRank": "MasseyOrdinalRank"
})
```

```python
team_features_men = team_features_men.merge(
    latest_massey_men,
    on=["Season", "TeamID"],
    how="left"
)
```

### Women's Massey ratings (computed from regular season results) @Jason

```python
import sys
sys.path.append("../scripts")
from massey_method import massey_method
```

```python
df_women_raw = pd.read_csv("../data/WRegularSeasonDetailedResults.csv")
```

```python
w_massey_list = []

for season in sorted(df_women_raw["Season"].unique()):
    season_df = df_women_raw[df_women_raw["Season"] == season].copy()
    ratings, sos, teams = massey_method(season_df)

    season_massey = pd.DataFrame({
        "Season": season,
        "TeamID": list(ratings.keys()),
        "WMasseyRating": list(ratings.values())
    })
    season_massey["WMasseySOS"] = season_massey["TeamID"].map(sos)
    w_massey_list.append(season_massey)
```

```python
w_massey_df = pd.concat(w_massey_list, ignore_index=True)
```

```python
team_features_women = team_features_women.merge(
    w_massey_df,
    on=["Season", "TeamID"],
    how="left"
)
```

### Replace infinite values

```python
numeric_cols_men = team_features_men.select_dtypes(include=[np.number]).columns
team_features_men[numeric_cols_men] = team_features_men[numeric_cols_men].replace(
    [np.inf, -np.inf], np.nan
)

numeric_cols_women = team_features_women.select_dtypes(include=[np.number]).columns
team_features_women[numeric_cols_women] = team_features_women[numeric_cols_women].replace(
    [np.inf, -np.inf], np.nan
)
```
