## Rebound and margin features
The same calculations are applied to both the men's and women's datasets.

### Features included

#### Total Rebounds @Dan
`TRB = OR + DR`

#### Rebound Percentage @Dan
`REB% = REB / (REB + OppREB)`

#### Offensive Rebound Percentage @Dan
`OREB% = OREB / (OREB + OppDREB)`

#### Defensive Rebound Percentage @Dan
`DREB% = DREB / (DREB + OppOREB)`

#### Turnover Margin @Gabe
`TurnoverMargin = OpponentTurnovers - TeamTurnovers`

#### Rebound Margin @Gabe
`ReboundMargin = TeamTotalRebounds - OpponentTotalRebounds`

```python
def add_rebound_margin_features(df):
    df = df.copy()

    df["W_TotalRebounds"] = df["W_OffensiveRebounds"] + df["W_DefensiveRebounds"]
    df["L_TotalRebounds"] = df["L_OffensiveRebounds"] + df["L_DefensiveRebounds"]

    df["W_ReboundPct"] = df["W_TotalRebounds"] / (
            df["W_TotalRebounds"] + df["L_TotalRebounds"]
    )
    df["L_ReboundPct"] = df["L_TotalRebounds"] / (
            df["W_TotalRebounds"] + df["L_TotalRebounds"]
    )

    df["W_OffensiveReboundPct"] = df["W_OffensiveRebounds"] / (
            df["W_OffensiveRebounds"] + df["L_DefensiveRebounds"]
    )
    df["L_OffensiveReboundPct"] = df["L_OffensiveRebounds"] / (
            df["L_OffensiveRebounds"] + df["W_DefensiveRebounds"]
    )

    df["W_DefensiveReboundPct"] = df["W_DefensiveRebounds"] / (
            df["W_DefensiveRebounds"] + df["L_OffensiveRebounds"]
    )
    df["L_DefensiveReboundPct"] = df["L_DefensiveRebounds"] / (
            df["L_DefensiveRebounds"] + df["W_OffensiveRebounds"]
    )

    df["W_TurnoverMargin"] = df["L_Turnovers"] - df["W_Turnovers"]
    df["L_TurnoverMargin"] = df["W_Turnovers"] - df["L_Turnovers"]

    df["W_ReboundMargin"] = df["W_TotalRebounds"] - df["L_TotalRebounds"]
    df["L_ReboundMargin"] = df["L_TotalRebounds"] - df["W_TotalRebounds"]

    return df
```

## Advanced team game statistics

This section applies the same game-level feature calculations to both the men's and women's datasets after merging winner and loser rows into one long-format dataframe.

The goal here is not to change the structure of the notebook, but to avoid repeating the same calculations twice.

### Stats included in this function

#### Field Goal Percentage @Gabe
`FG% = FGM / FGA`

#### 3-Point Percentage @Gabe
`3P% = 3PM / 3PA`

#### Free Throw Percentage @Gabe
`FT% = FTM / FTA`

#### Turnover Percentage @Gabe
`TO% = TO / (FGA + 0.44 * FTA + TO)`

#### Assist-to-Turnover Ratio @Gabe
`AST/TO = AST / TO`

#### Total Game Points @Gabe
`GameTotalPoints = PointsFor + PointsAgainst`

#### Possessions @Hafsa
`Poss = FGA - OR + TO + 0.44 * FTA`

#### Offensive Efficiency @Hafsa
`OffEfficiency = PointsFor / Possessions`

#### Defensive Efficiency @Hafsa
`DefEfficiency = PointsAgainst / Possessions`

#### Net Rating @Hafsa
`NetRating = OffEfficiency - DefEfficiency`

#### Games Played @Gabe
`GamesPlayed = total number of games played by team in a season`

#### Wins @Gabe
`Wins = total number of wins by team in a season`

#### Win Percentage @Gabe
`Win% = Wins / GamesPlayed`

#### Losses @Gabe
`Losses = GamesPlayed - Wins`

#### Loss Percentage @Gabe
`Loss% = Losses / GamesPlayed`

```python
def add_advanced_stats(df):
    df = df.copy()

    df["FieldGoalsPct"] = df["FieldGoalsMade"] / df["FieldGoalsAttempted"]
    df["ThreePointsMadePct"] = df["ThreePointsMade"] / df["ThreePointsAttempted"]
    df["FreeThrowPct"] = df["FreeThrowsMade"] / df["FreeThrowsAttempted"]

    df["TurnoverPct"] = df["Turnovers"] / (
            df["FieldGoalsAttempted"] + 0.44 * df["FreeThrowsAttempted"] + df["Turnovers"]
    )

    df["AssistTurnoverRto"] = df["Assists"] / df["Turnovers"]
    df["GameTotalPoints"] = df["PointsAgainst"] + df["PointsFor"]

    df["Possessions"] = (
            df["FieldGoalsAttempted"]
            - df["OffensiveRebounds"]
            + df["Turnovers"]
            + 0.44 * df["FreeThrowsAttempted"]
    )

    df["OffEfficiency"] = df["PointsFor"] / df["Possessions"]
    df["DefEfficiency"] = df["PointsAgainst"] / df["Possessions"]
    df["NetRating"] = df["OffEfficiency"] - df["DefEfficiency"]

    df["GamesPlayed"] = df.groupby(["Season", "TeamID"])["Win"].transform("count")
    df["Wins"] = df.groupby(["Season", "TeamID"])["Win"].transform("sum")
    df["WinPct"] = df["Wins"] / df["GamesPlayed"]

    df["Losses"] = df["GamesPlayed"] - df["Wins"]
    df["LossPct"] = df["Losses"] / df["GamesPlayed"]

    return df
```

```python
games_men = add_advanced_stats(games_men)
games_women = add_advanced_stats(games_women)
```

## Seeds

```python
m_seeds = pd.read_csv("../data/MNCAATourneySeeds.csv")
w_seeds = pd.read_csv("../data/WNCAATourneySeeds.csv")
```

```python
def extract_seed_number(seed_value):
    if pd.isna(seed_value):
        return np.nan

    digits = "".join(ch for ch in str(seed_value) if ch.isdigit())
    return int(digits) if digits else np.nan
```

```python
m_seeds["SeedNum"] = m_seeds["Seed"].apply(extract_seed_number)
m_seeds["HasTournamentSeed"] = 1

w_seeds["SeedNum"] = w_seeds["Seed"].apply(extract_seed_number)
w_seeds["HasTournamentSeed"] = 1
```
