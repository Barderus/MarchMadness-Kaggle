## Rebound and margin features
The same calculations are applied to both the men's and women's datasets.

### Features included

#### Total Rebounds @Dan
`TRB = OR + DR`

#### Rebound Percentage @Dan
`REB% = REB / (REB + OppREB)`

#### Offensive Rebound Percentage @Dan
`OREB% = OREB / (OREB + OppDREB)`

#### Defensive Rebound Percentage @Dan
`DREB% = DREB / (DREB + OppOREB)`

#### Turnover Margin @Gabe
`TurnoverMargin = OpponentTurnovers - TeamTurnovers`

#### Rebound Margin @Gabe
`ReboundMargin = TeamTotalRebounds - OpponentTotalRebounds`

```python
def safe_divide(numerator, denominator):
    return np.where(denominator != 0, numerator / denominator, 0)
```

```python
def add_rebound_margin_features(df):
    df = df.copy()

    df["W_TotalRebounds"] = df["W_OffensiveRebounds"] + df["W_DefensiveRebounds"]
    df["L_TotalRebounds"] = df["L_OffensiveRebounds"] + df["L_DefensiveRebounds"]

    df["W_ReboundPct"] = safe_divide(
        df["W_TotalRebounds"],
        df["W_TotalRebounds"] + df["L_TotalRebounds"]
    )
    df["L_ReboundPct"] = safe_divide(
        df["L_TotalRebounds"],
        df["W_TotalRebounds"] + df["L_TotalRebounds"]
    )

    df["W_OffensiveReboundPct"] = safe_divide(
        df["W_OffensiveRebounds"],
        df["W_OffensiveRebounds"] + df["L_DefensiveRebounds"]
    )
    df["L_OffensiveReboundPct"] = safe_divide(
        df["L_OffensiveRebounds"],
        df["L_OffensiveRebounds"] + df["W_DefensiveRebounds"]
    )

    df["W_DefensiveReboundPct"] = safe_divide(
        df["W_DefensiveRebounds"],
        df["W_DefensiveRebounds"] + df["L_OffensiveRebounds"]
    )
    df["L_DefensiveReboundPct"] = safe_divide(
        df["L_DefensiveRebounds"],
        df["L_DefensiveRebounds"] + df["W_OffensiveRebounds"]
    )

    df["W_TurnoverMargin"] = df["L_Turnovers"] - df["W_Turnovers"]
    df["L_TurnoverMargin"] = df["W_Turnovers"] - df["L_Turnovers"]

    df["W_ReboundMargin"] = df["W_TotalRebounds"] - df["L_TotalRebounds"]
    df["L_ReboundMargin"] = df["L_TotalRebounds"] - df["W_TotalRebounds"]

    return df
```

```python
df_men = add_rebound_margin_features(df_men)
df_women = add_rebound_margin_features(df_women)
```

Convert wide game results into long team game format

```python
def build_team_game_rows(df, team_prefix, opp_prefix, win_value):
    return pd.DataFrame({
        "Season": df["Season"],
        "DayNum": df["DayNum"],
        "NumOT": df["NumOT"],
        "TeamID": df[f"{team_prefix}TeamID"],
        "OpponentID": df[f"{opp_prefix}TeamID"],
        "Win": win_value,
        "PointsFor": df[f"{team_prefix}Score"],
        "PointsAgainst": df[f"{opp_prefix}Score"],
        "FieldGoalsMade": df[f"{team_prefix}_FieldGoalsMade"],
        "FieldGoalsAttempted": df[f"{team_prefix}_FieldGoalsAttempted"],
        "ThreePointsMade": df[f"{team_prefix}_ThreePointsMade"],
        "ThreePointsAttempted": df[f"{team_prefix}_ThreePointsAttempted"],
        "FreeThrowsMade": df[f"{team_prefix}_FreeThrowsMade"],
        "FreeThrowsAttempted": df[f"{team_prefix}_FreeThrowsAttempted"],
        "OffensiveRebounds": df[f"{team_prefix}_OffensiveRebounds"],
        "DefensiveRebounds": df[f"{team_prefix}_DefensiveRebounds"],
        "Assists": df[f"{team_prefix}_Assists"],
        "Turnovers": df[f"{team_prefix}_Turnovers"],
        "Steals": df[f"{team_prefix}_Steals"],
        "Blocks": df[f"{team_prefix}_Blocks"],
        "PersonalFouls": df[f"{team_prefix}_PersonalFouls"],
        "TotalRebounds": df[f"{team_prefix}_TotalRebounds"],
        "ReboundPct": df[f"{team_prefix}_ReboundPct"],
        "OffensiveReboundPct": df[f"{team_prefix}_OffensiveReboundPct"],
        "DefensiveReboundPct": df[f"{team_prefix}_DefensiveReboundPct"],
        "OpponentTurnovers": df[f"{opp_prefix}_Turnovers"],
        "OpponentTotalRebounds": df[f"{opp_prefix}_TotalRebounds"],
        "TurnoverMargin": df[f"{team_prefix}_TurnoverMargin"],
        "ReboundMargin": df[f"{team_prefix}_ReboundMargin"],
    })
```

```python
def build_games_long(df):
    winner_df = build_team_game_rows(df, team_prefix="W", opp_prefix="L", win_value=1)
    loser_df = build_team_game_rows(df, team_prefix="L", opp_prefix="W", win_value=0)

    winner_df["MarginScore"] = winner_df["PointsFor"] - winner_df["PointsAgainst"]
    loser_df["MarginScore"] = loser_df["PointsFor"] - loser_df["PointsAgainst"]

    return pd.concat([winner_df, loser_df], ignore_index=True)
```

```python
games_men = build_games_long(df_men)
games_women = build_games_long(df_women)
```

## Advanced team game statistics

This section applies the same game-level feature calculations to both the men's and women's datasets after merging winner and loser rows into one long-format dataframe.

The goal here is not to change the structure of the notebook, but to avoid repeating the same calculations twice.

### Stats included in this function

#### Field Goal Percentage @Gabe
`FG% = FGM / FGA`

#### 3-Point Percentage @Gabe
`3P% = 3PM / 3PA`

#### Free Throw Percentage @Gabe
`FT% = FTM / FTA`

#### Turnover Percentage @Gabe
`TO% = TO / (FGA + 0.44 * FTA + TO)`

#### Assist-to-Turnover Ratio @Gabe
`AST/TO = AST / TO`

#### Total Game Points @Gabe
`GameTotalPoints = PointsFor + PointsAgainst`

#### Possessions @Hafsa
`Poss = FGA - OR + TO + 0.44 * FTA`

#### Offensive Efficiency @Hafsa
`OffEfficiency = PointsFor / Possessions`

#### Defensive Efficiency @Hafsa
`DefEfficiency = PointsAgainst / Possessions`

#### Net Rating @Hafsa
`NetRating = OffEfficiency - DefEfficiency`

#### Games Played @Gabe
`GamesPlayed = total number of games played by team in a season`

#### Wins @Gabe
`Wins = total number of wins by team in a season`

#### Win Percentage @Gabe
`Win% = Wins / GamesPlayed`

#### Losses @Gabe
`Losses = GamesPlayed - Wins`

#### Loss Percentage @Gabe
`Loss% = Losses / GamesPlayed`

```python
def add_advanced_stats(df):
    df = df.copy()

    df["FieldGoalsPct"] = safe_divide(df["FieldGoalsMade"], df["FieldGoalsAttempted"])
    df["ThreePointsPct"] = safe_divide(df["ThreePointsMade"], df["ThreePointsAttempted"])
    df["FreeThrowPct"] = safe_divide(df["FreeThrowsMade"], df["FreeThrowsAttempted"])

    df["TurnoverPct"] = safe_divide(
        df["Turnovers"],
        df["FieldGoalsAttempted"] + 0.44 * df["FreeThrowsAttempted"] + df["Turnovers"]
    )

    df["AssistTurnoverRatio"] = safe_divide(df["Assists"], df["Turnovers"])
    df["GameTotalPoints"] = df["PointsAgainst"] + df["PointsFor"]

    df["Possessions"] = (
            df["FieldGoalsAttempted"]
            - df["OffensiveRebounds"]
            + df["Turnovers"]
            + 0.44 * df["FreeThrowsAttempted"]
    )

    df["OffEfficiency"] = safe_divide(df["PointsFor"], df["Possessions"])
    df["DefEfficiency"] = safe_divide(df["PointsAgainst"], df["Possessions"])
    df["NetRating"] = df["OffEfficiency"] - df["DefEfficiency"]

    df["GamesPlayed"] = df.groupby(["Season", "TeamID"])["Win"].transform("count")
    df["Wins"] = df.groupby(["Season", "TeamID"])["Win"].transform("sum")
    df["WinPct"] = safe_divide(df["Wins"], df["GamesPlayed"])

    df["Losses"] = df["GamesPlayed"] - df["Wins"]
    df["LossPct"] = safe_divide(df["Losses"], df["GamesPlayed"])

    return df
```

```python
games_men = add_advanced_stats(games_men)
games_women = add_advanced_stats(games_women)
```
