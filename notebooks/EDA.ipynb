```python
import pandas as pd
import numpy as np
```

### Load files

```python
df_men = pd.read_csv("../data/MRegularSeasonDetailedResults.csv")
df_women = pd.read_csv("../data/WRegularSeasonDetailedResults.csv")
```

### Renaming files to increase readability

```python
headers_rename = {

    # Winner team stats
    "WFGM":  "W_FieldGoalsMade",
    "WFGA":  "W_FieldGoalsAttempted",
    "WFGM3": "W_ThreePointsMade",
    "WFGA3": "W_ThreePointsAttempted",
    "WFTM":  "W_FreeThrowsMade",
    "WFTA":  "W_FreeThrowsAttempted",
    "WOR":   "W_OffensiveRebounds",
    "WDR":   "W_DefensiveRebounds",
    "WAst":  "W_Assists",
    "WTO":   "W_Turnovers",
    "WStl":  "W_Steals",
    "WBlk":  "W_Blocks",
    "WPF":   "W_PersonalFouls",

    # Losing team stats
    "LFGM":  "L_FieldGoalsMade",
    "LFGA":  "L_FieldGoalsAttempted",
    "LFGM3": "L_ThreePointsMade",
    "LFGA3": "L_ThreePointsAttempted",
    "LFTM":  "L_FreeThrowsMade",
    "LFTA":  "L_FreeThrowsAttempted",
    "LOR":   "L_OffensiveRebounds",
    "LDR":   "L_DefensiveRebounds",
    "LAst":  "L_Assists",
    "LTO":   "L_Turnovers",
    "LStl":  "L_Steals",
    "LBlk":  "L_Blocks",
    "LPF":   "L_PersonalFouls"
}
```

```python
df_women.rename(columns = headers_rename, inplace = True)
df_women.head(5)
```

#### Total Rebounds

```python
df_men["W_TotalRebounds"] = df_men["W_OffensiveRebounds"] + df_men["W_DefensiveRebounds"]
df_men["L_TotalRebounds"] = df_men["L_OffensiveRebounds"] + df_men["L_DefensiveRebounds"]
```

#### Rebound % for winning and losing teams

```python
df_men["W_ReboundPct"] = df_men["W_TotalRebounds"] / (df_men["W_TotalRebounds"] + df_men["L_TotalRebounds"])
```

```python
df_men["L_ReboundPct"] = df_men["L_TotalRebounds"] / (df_men["W_TotalRebounds"] + df_men["L_TotalRebounds"])
```

#### Offensive Rebound %

```python
df_men["W_OffensiveReboundPct"] = df_men["W_OffensiveRebounds"] / (df_men["W_OffensiveRebounds"] + df_men["L_DefensiveRebounds"])
```

```python
df_men["L_OffensiveReboundPct"] = df_men["L_OffensiveRebounds"] / (df_men["L_OffensiveRebounds"] + df_men["W_DefensiveRebounds"])
```

#### Defensive Rebound %

```python
df_men["W_DefensiveReboundPct"] = df_men["W_DefensiveRebounds"] / (df_men["W_DefensiveRebounds"] + df_men["L_OffensiveRebounds"])
```

```python
df_men["L_DefensiveReboundPct"] = df_men["L_DefensiveRebounds"] / (df_men["L_DefensiveRebounds"] + df_men["W_OffensiveRebounds"])
```

```python
df_men["W_TurnoverMargin"] = df_men["L_Turnovers"] - df_men["W_Turnovers"]
df_men["L_TurnoverMargin"] = df_men["W_Turnovers"] - df_men["L_Turnovers"]
```

```python
df_men["W_ReboundMargin"] = df_men["W_TotalRebounds"] - df_men["L_TotalRebounds"]
df_men["L_ReboundMargin"] = df_men["L_TotalRebounds"] - df_men["W_TotalRebounds"]
```

### Let's separate the df between winners and losers stats

```python
winner_df = pd.DataFrame({
    "Season": df_men["Season"],
    "DayNum": df_men["DayNum"],
    "NumOT": df_men["NumOT"],
    "TeamID": df_men["WTeamID"],
    "OpponentID": df_men["LTeamID"],
    "Win": 1,
    "PointsFor": df_men["WScore"],
    "PointsAgainst": df_men["LScore"],
    "FieldGoalsMade": df_men["W_FieldGoalsMade"],
    "FieldGoalsAttempted": df_men["W_FieldGoalsAttempted"],
    "ThreePointsMade": df_men["W_ThreePointsMade"],
    "ThreePointsAttempted": df_men["W_ThreePointsAttempted"],
    "FreeThrowsMade": df_men["W_FreeThrowsMade"],
    "FreeThrowsAttempted": df_men["W_FreeThrowsAttempted"],
    "OffensiveRebounds": df_men["W_OffensiveRebounds"],
    "DefensiveRebounds": df_men["W_DefensiveRebounds"],
    "Assists": df_men["W_Assists"],
    "Turnovers": df_men["W_Turnovers"],
    "Steals": df_men["W_Steals"],
    "Blocks": df_men["W_Blocks"],
    "PersonalFouls": df_men["W_PersonalFouls"],
    "TotalRebounds": df_men["W_TotalRebounds"],
    "ReboundPct": df_men["W_ReboundPct"],
    "OffensiveReboundPct": df_men["W_OffensiveReboundPct"],
    "DefensiveReboundPct": df_men["W_DefensiveReboundPct"],
    "OpponentTurnovers": df_men["L_Turnovers"],
    "OpponentTotalRebounds": df_men["L_TotalRebounds"],
    "TurnoverMargin": df_men["W_TurnoverMargin"],
    "ReboundMargin": df_men["W_ReboundMargin"],
})
winner_df.head(5)
```

```python
loser_df = pd.DataFrame({
    "Season": df_men["Season"],
    "DayNum": df_men["DayNum"],
    "NumOT": df_men["NumOT"],
    "TeamID": df_men["LTeamID"],
    "OpponentID": df_men["WTeamID"],
    "Win": 0,
    "PointsFor": df_men["LScore"],
    "PointsAgainst": df_men["WScore"],
    "FieldGoalsMade": df_men["L_FieldGoalsMade"],
    "FieldGoalsAttempted": df_men["L_FieldGoalsAttempted"],
    "ThreePointsMade": df_men["L_ThreePointsMade"],
    "ThreePointsAttempted": df_men["L_ThreePointsAttempted"],
    "FreeThrowsMade": df_men["L_FreeThrowsMade"],
    "FreeThrowsAttempted": df_men["L_FreeThrowsAttempted"],
    "OffensiveRebounds": df_men["L_OffensiveRebounds"],
    "DefensiveRebounds": df_men["L_DefensiveRebounds"],
    "Assists": df_men["L_Assists"],
    "Turnovers": df_men["L_Turnovers"],
    "Steals": df_men["L_Steals"],
    "Blocks": df_men["L_Blocks"],
    "PersonalFouls": df_men["L_PersonalFouls"],
    "TotalRebounds": df_men["L_TotalRebounds"],
    "ReboundPct": df_men["L_ReboundPct"],
    "OffensiveReboundPct": df_men["L_OffensiveReboundPct"],
    "DefensiveReboundPct": df_men["L_DefensiveReboundPct"],
    "OpponentTurnovers": df_men["W_Turnovers"],
    "OpponentTotalRebounds": df_men["W_TotalRebounds"],
    "TurnoverMargin": df_men["L_TurnoverMargin"],
    "ReboundMargin": df_men["L_ReboundMargin"],
})
loser_df.head(5)
```

# Possessions @Hafsa
`Poss = FGA - OR + TO + 0.44 * FTA`

```python
games_men["Possessions"] = (
    (games_men["FieldGoalsAttempted"]- games_men["OffensiveRebounds"] + games_men["Turnovers"] + 0.44 * games_men["FreeThrowsAttempted"]
     ))

games_men["Possessions"].head()
```

# Offensive efficiency @Hafsa
`offRating = Points / Possesion`

```python
games_men["OffEfficiency"]= games_men["PointsFor"] / games_men["Possessions"]
games_men["OffEfficiency"].head()
```

# Defensive efficiency @Hafsa
defRating = OppPoints / Possessions

```python
games_men["DefEfficiency"]= games_men["PointsAgainst"] / games_men["Possessions"]
games_men["DefEfficiency"].head()
```

# Net efficiency @ Hafsa
`NetRating = offRating - defRating`

```python
games_men["NetRating"] = games_men["OffEfficiency"] - games_men["DefEfficiency"]
games_men["NetRating"].head()
```

# Team win percentage @Gabe
`wins = wins / games`

```python
games_men["GamesPlayed"] = games_men.groupby(["Season", "TeamID"])["Win"].transform("count")
```

```python
games_men["Wins"] = games_men.groupby(["Season", "TeamID"])["Win"].transform("sum")
games_men["WinPct"] = games_men["Wins"] / games_men["GamesPlayed"]
games_men[["Season","TeamID","Wins","GamesPlayed","WinPct"]].head()
```

```python
games_men["Losses"] = games_men["GamesPlayed"] - games_men["Wins"]
games_men["LossPct"] = games_men["Losses"] / games_men["GamesPlayed"]
games_men[["Season","TeamID","Losses","GamesPlayed","LossPct"]].head()
```
